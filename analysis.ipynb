# Insights

• Employee attrition is lower compared to employee retention, indicating overall workforce stability.

• Employees aged between 25–35 years show higher attrition tendencies compared to other age groups.

• Lower salary employees are more likely to leave the organization, suggesting compensation influences retention.

• Employees working overtime exhibit higher attrition patterns, indicating workload impacts employee turnover.

• Employees with shorter tenure (1–5 years) are more vulnerable to attrition compared to long-tenured employees.

• Research & Development and Sales departments show relatively higher attrition counts.

• Work-life balance, compensation, career growth opportunities, and employee experience appear to significantly influence employee retention.

• Employee attrition is influenced by multiple interconnected factors rather than a single variable.
